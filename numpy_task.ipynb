```python
import numpy as np

# Creating 5 samples with 3 features
X = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90],
    [15, 25, 35],
    [45, 55, 65]
])

# Checking the number of rows and columns
print("Shape:", X.shape)
```

```python
# Finding the average value of each feature
feature_mean = np.mean(X, axis=0)

print("Feature means:", feature_mean)
```

```python
# Selecting rows where the first feature is greater than 50
filtered_rows = X[X[:, 0] > 50]

print(filtered_rows)
```

```python
# Finding the mean and standard deviation of each feature
mean = np.mean(X, axis=0)
std = np.std(X, axis=0)

# Standardizing each feature column
X_standardized = (X - mean) / std

print("Standardized data:")
print(X_standardized)
```

```python
# Setting a seed so we get the same random values every time
np.random.seed(42)

# Creating a random matrix with 10 rows and 4 columns
random_matrix = np.random.rand(10, 4)

print(random_matrix)
```

```python
X = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

weights = np.array([0.2, 0.3, 0.5])

# Multiplying the features with their corresponding weights
result = X @ weights

print(result)
```

```python
X = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

bias = np.array([1, 2, 3])

# Adding the bias to each row
result = X + bias

print(result)
```

```python
y_true = np.array([10, 20, 30, 40])
y_pred = np.array([12, 18, 33, 37])

# Finding the difference between actual and predicted values
error = y_true - y_pred

# Calculating the mean squared error
mse = np.mean(error ** 2)

print("MSE:", mse)
```

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

# Multiplying corresponding elements
elementwise_result = A * B

# Performing matrix multiplication
matrix_result = A @ B

print("A * B:")
print(elementwise_result)

print("A @ B:")
print(matrix_result)
```

```python
import numpy as np

data = np.array([10, 20, 30, 40, 50], dtype=float)

# Scaling the values between 0 and 1
normalized = (data - np.min(data)) / (np.max(data) - np.min(data))

# Standardizing the values using mean and standard deviation
standardized = (data - np.mean(data)) / np.std(data)

print("Original:", data)
print("Normalized:", normalized)
print("Standardized:", standardized)

# Checking the mean and standard deviation after standardization
print("Standardized mean:", np.mean(standardized))
print("Standardized std:", np.std(standardized))
```
